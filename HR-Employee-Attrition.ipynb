```python
# Attrition Flag
```

```python
df["Attrition_Flag"] = np.where(
    df["Attrition"] == "Yes",
    1,
    0
)
```

```python
# Age Group
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 29, 39, np.inf],
    labels=["18-29", "30-39", "40+"]
)
```

```python
df["Age_Group"].value_counts().sort_index()
```

```python
total_employees = len(df)

employees_left = df["Attrition_Flag"].sum()

attrition_rate = df["Attrition_Flag"].mean() * 100

average_age = df["Age"].mean()

average_income = df["MonthlyIncome"].mean()

print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print(f"Attrition Rate: {attrition_rate:.2f}%")
print(f"Average Age: {average_age:.2f}")
print(f"Average Monthly Income: ₹{average_income:,.2f}")
```

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

department_attrition
```

```python
#Attrition by Department
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.tight_layout()
plt.show()
```

```python
#Attrition by Job Role
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="JobRole",
    hue="Attrition"
)

plt.title("Attrition by Job Role")
plt.xlabel("Employee Count")
plt.ylabel("Job Role")

plt.tight_layout()
plt.show()
```

```python
#Attrition by Age Group
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Age_Group",
    hue="Attrition",
    order=["18-29", "30-39", "40+"]
)

plt.title("Attrition by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Employee Count")

plt.tight_layout()
plt.show()
```

```python
#Attrition by Overtime
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition"
)

plt.title("Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Employee Count")

plt.tight_layout()
plt.show()
```

```python
#Attrition by Job Satisfaction
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="Attrition"
)

plt.title("Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Employee Count")

plt.tight_layout()
plt.show()
```

```python
attrition_by_years = (
    df.groupby("YearsAtCompany")["Attrition_Flag"]
    .mean()
    .mul(100)
    .reset_index()
)
```

```python
#Attrition Rate by Years at Company
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    data=attrition_by_years,
    x="YearsAtCompany",
    y="Attrition_Flag",
    marker="o"
)

plt.title("Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()
```

```python
#Monthly Income vs Attrition
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income vs Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.show()
```
